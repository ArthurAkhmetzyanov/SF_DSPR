```python
data = pd.read_excel('data_ford_price.xlsx') 
```

```python
#Предобработка данных
data = data[['price','year', 'cylinders', 'odometer', 'lat', 'long', 'weather']]
data.dropna(inplace = True)
```

```python
# Разделение данных на признаки и целевую переменную
y = data['price']
x = data.drop(columns='price')

# Разделение на обучающую и тестовую выборки
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

estimator = LinearRegression()
selector = RFE(estimator, n_features_to_select=3, step=1)
selector = selector.fit(X_train, y_train)
 
selector.get_feature_names_out()
```

```python
#Обучаем модель на трех столбцах выбраных RFE
rfe_data = data[['year', 'cylinders', 'lat', 'price']]

# Разделение данных на признаки и целевую переменную
y = rfe_data['price']
x = rfe_data.drop(columns='price')

# Разделение на обучающую и тестовую выборки
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

# Обучение модели
model = LinearRegression()
model.fit(X_train, y_train)
y_predicted = model.predict(X_test)

# Вычисление MAE 
mae = mean_absolute_error(y_test, y_predicted)
print('MAE: %.3f' % mae)
```

```python
# Разделение данных на признаки и целевую переменную
y = data['price']
x = data.drop(columns='price')

# Разделение на обучающую и тестовую выборки
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

selector = SelectKBest(f_regression, k=3)
selector.fit(X_train, y_train)
 
selector.get_feature_names_out()
```

```python
#Обучаем модель на трех столбцах выбраных SelectKBest
skb_data = data[['year', 'cylinders', 'odometer', 'price']]

# Разделение данных на признаки и целевую переменную
y = skb_data['price']
x = skb_data.drop(columns='price')

# Разделение на обучающую и тестовую выборки
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

# Обучение модели
model = LinearRegression()
model.fit(X_train, y_train)
y_predicted = model.predict(X_test)

# Вычисление MAE 
mae = mean_absolute_error(y_test, y_predicted)
print('MAE: %.3f' % mae)
```
