```python
import re
```

```python
s = '2020-12-14'
match = re.search(r'(\d\d\d\d)-(\d\d)-(\d\d)', s)
if match:
    print(f'Matched items: {match.groups(1)}')
else:
    print(f'No match found is {s}')
```

## <center> Глава 5. Управление выполнением

Оператор присванивания `(:=)` - его называют мажороным оператором

```python
s = '2020-12-14'
if match := re.search(r'(\d\d\d\d)-(\d\d)-(\d\d)', s):
    print(f'Matched items: {match.groups(1)}')
else:
    print(f'No match found is {s}')
```
