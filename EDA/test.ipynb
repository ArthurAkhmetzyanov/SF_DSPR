```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Генерация нормально распределённых данных
data_normal = np.random.normal(loc=0, scale=1, size=1000)

# Добавление выбросов
data_outliers = np.concatenate([data_normal, np.random.normal(loc=5, scale=0.5, size=50)])

# Визуализация
fig, ax = plt.subplots(1, 2, figsize=(14, 5))

# Возможная проверка на содержимое
if data_normal.size > 0 and data_outliers.size > 0:
    sns.histplot(data_normal, bins=30, kde=True, ax=ax[0])
    ax[0].set_title('Нормальное Распределение')
    ax[0].set_xlabel('Значение')
    ax[0].set_ylabel('Частота')

    sns.histplot(data_outliers, bins=30, kde=True, ax=ax[1])
    ax[1].set_title('Распределение с Выбросами')
    ax[1].set_xlabel('Значение')
    ax[1].set_ylabel('Частота')

    plt.show()
else:
    print("Данные отсутствуют или ошибка в данных.")
plt.show()
```
