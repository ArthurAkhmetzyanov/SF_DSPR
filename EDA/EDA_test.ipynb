```python
df = pd.read_csv('data/new-site.csv', sep='\t')
```

```python
df
```

```python
sum_dwell_time = df.groupby('site')['dwell-time'].sum()
```

```python
print("\nСуммарное значение dwell-time для каждого site:")
print(sum_dwell_time)
```

```python
mean_dwell_time = df.groupby('site')['dwell-time'].mean()
print("Среднее значение dwell-time для каждого site:")
print(round(mean_dwell_time, 0))
```

```python
dwell_time_groups = df.groupby('site')['dwell-time'].apply(list).to_dict()

# Получение списков значений dwell-time для каждого site
dwell_time_site_0 = dwell_time_groups.get(0, [])
dwell_time_site_1 = dwell_time_groups.get(1, [])

print("dwell-time для site 0:", dwell_time_site_0)
print("dwell-time для site 1:", dwell_time_site_1)
```

```python
import scipy.stats as stats
```

```python
alpha = 0.05

_, p = stats.shapiro(dwell_time_site_0) # группа А

print('p-value = %.3f' % (p))

if p <= alpha:
    print('Распределение не нормальное')
else:
    print('Распределение нормальное')
```

```python
alpha = 0.05

_, p = stats.shapiro(dwell_time_site_1) # группа B

print('p-value = %.3f' % (p))

if p <= alpha:
    print('Распределение не нормальное')
else:
    print('Распределение нормальное')
```

```python
# задаём уровень значимости
alpha = 0.05 


# проводим тест
_, p = stats.mannwhitneyu(dwell_time_site_0, dwell_time_site_1)

print('p-value = {:.3f}'.format(p))

# интерпретируем результат
if p <= alpha:
    print('p-значение меньше, чем заданный уровень значимости {:.2f}. Отвергаем нулевую гипотезу.'.format(alpha))
else:
    print('p-значение больше, чем заданный уровень значимости {:.2f}. У нас нет оснований отвергнуть нулевую гипотезу.'.format(alpha))
```
